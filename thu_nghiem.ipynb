```python
import os
from tqdm import tqdm
from langchain_community.document_loaders import UnstructuredPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Cấu hình Hugging Face Token (dùng để tải model nhanh và ổn định)
os.environ["HF_TOKEN"] = "hf_...pWNf" # Dán mã Token đầy đủ của bạn vào đây

# 2. Load tài liệu với chế độ hi_res và tiếng Việt
file_path = "SỔ TAY SV - Bách Khoa - NĂM HỌC 2025-2026.pdf"
print("Đang đọc tài liệu (quá trình này có thể mất chút thời gian)...")
loader = UnstructuredPDFLoader(
    file_path,
    strategy="hi_res", 
    mode="elements",
    languages=["vie"]
)

# Sử dụng tqdm để theo dõi tiến trình load
documents = loader.load()
print(f"Đã load xong {len(documents)} phần tử.")

# 3. Cấu hình chia nhỏ tài liệu
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=250,
    separators=["\n\n", "\n", "PHẦN", "Điều", "Bước", ".", " ", ""]
)

print("Đang chia nhỏ tài liệu...")
chunks = text_splitter.split_documents(documents)

# 4. Thêm Metadata với thanh tiến trình
for i, chunk in tqdm(enumerate(chunks), total=len(chunks), desc="Đang xử lý Metadata"):
    chunk.metadata["source_name"] = "So_tay_sinh_vien"
    chunk.metadata["chunk_index"] = i

# 5. Kiểm tra kết quả
print(f"\n--- Đã hoàn tất chia nhỏ tài liệu ---")
print(f"Tổng số đoạn (chunks) tạo ra: {len(chunks)}")
print(f"\n--- Xem thử đoạn đầu tiên ---")
print(f"Metadata: {chunks[0].metadata}")
print(f"Nội dung: {chunks[0].page_content[:500]}...")
```

```python
# In thử đoạn số 500 (khoảng giữa tài liệu) để kiểm tra chất lượng OCR
if len(chunks) > 500:
    print(f"--- Kiểm tra đoạn số 500 ---")
    print(chunks[500].page_content)
else:
    print("Tổng số đoạn ít hơn 500, không thể test đoạn này.")
```
